```python
import pandas as pd
```

```python
COUNTRY_CODE = "UKR"
NAME_TYPES = ["given_name", "middle_name", "family_name"]
```

```python
name_df = pd.read_csv(f"../../../data/name_parser/input/{COUNTRY_CODE}.csv", sep=";", usecols=NAME_TYPES)
name_df.head()
```

```python
name_df.shape
```

```python
name_dfs = []
for name in NAME_TYPES:
    name_df = name_df[name].rename("name").to_frame()
    name_df["type"] = name
    name_dfs.append(name_df)
```

```python
name_df = pd.concat(name_dfs)
name_df.shape
```

```python
name_df["name"] = name_df["name"].str.replace(r"[^-\w' ]", "", regex=True)
name_df.head()
```

```python
name_df["name"] = name_df["name"].str.strip().str.title()
name_df = name_df.dropna()
name_df = name_df.drop_duplicates()
name_df.shape
```

```python
name_df
```

```python
name_df.to_csv(f"../../../data/name_parser/tmp/{COUNTRY_CODE}.csv", index=False)
```
